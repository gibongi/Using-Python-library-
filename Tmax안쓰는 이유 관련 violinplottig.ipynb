```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. CSV 파일 불러오기 ('DMSP_F16_SOLARmax_Kpge2lt4_ele_v02.csv' 사용)
df = pd.read_csv('DMSP_F16_SOLARmax_Kpge2lt4_ele_v02.csv')

# T_e_q3 Violin Plot 그리기
sns.violinplot( y='T_e_q3', data=df)
plt.title('T_e_q3 Distribution')
plt.show()
```

```python
# T_e_max Violin Plot 그리기
sns.violinplot( y='T_e_max', data=df)
plt.title('T_e_max Distribution')
plt.show()
```

```python
# '데이터 형태 변환 (Wide to Long)'
# 두 열을 하나로 합쳐서 시각화하기 위해 pandas의 melt 기능을 사용
df_melted = df.melt(value_vars=['T_e_q3', 'T_e_max'], var_name='T_e_q3_and_T_e_max', value_name='value')

# 3. Violin Plot 그리기
plt.figure(figsize=(10, 6))
sns.violinplot(x='T_e_q3_and_T_e_max', y='value', data=df_melted)

plt.title('Comparison of Two Columns')
plt.show()
```

```python
#T_e_err_max Violin Plot 그리기
sns.violinplot( y='T_e_err_max', data=df)
plt.title('T_e_err_max Distribution')
plt.show()
```

```python
# T_e_err_q3 Violin Plot 그리기
sns.violinplot( y='T_e_err_q3', data=df)
plt.title('T_e_err_q3 Distribution')
plt.show()
```

```python
# '데이터 형태 변환 (Wide to Long)'
# 두 열을 하나로 합쳐서 시각화하기 위해 pandas의 melt 기능을 사용
df_melted = df.melt(value_vars=['T_e_err_q3', 'T_e_err_max'], var_name='T_e_err_q3_and_T_e_err_max', value_name='value')

# 3. Violin Plot 그리기
plt.figure(figsize=(10, 6))
sns.violinplot(x='T_e_err_q3_and_T_e_err_max', y='value', data=df_melted)

plt.title('Comparison of Two Columns')
plt.show()
```

```python
# x축에 그룹을 넣어서 범주를 쪼개어서  그리기 (범주화한다 / 일정한 bin값 8에 따른 value 분포도)
df['Group'] = pd.cut(df['T_e_max'], bins=8)
plt.figure(figsize=(12, 6))
sns.violinplot(x='Group', y='T_e_max', data=df)
plt.xticks(rotation=45)
plt.show()

df['Group'] = pd.cut(df['T_e_q3'], bins=8)
plt.figure(figsize=(12, 6))
sns.violinplot(x='Group', y='T_e_q3', data=df)
plt.xticks(rotation=45)
plt.show()
```
