```python
import numpy as np
import matplotlib.pyplot as plt


## 
# 1. 자기 레이놀즈 수 (Rm) 범위 설정 (100에서 10^9까지)
Rm = np.logspace(2, 9, 500)

# 2. 각 모델의 재결합 속도(MA) 정의
# Sweet-Parker: MA ~ Rm^(-0.5)
ma_sp = Rm**(-0.5)

# Petschek: MA ~ (pi / 8) / ln(Rm)
ma_petschek = (np.pi / 8) / np.log(Rm)

# 3. 그래프 그리기
plt.figure(figsize=(10, 6))

# 데이터 플롯 (로그-로그 스케일 설정)
plt.loglog(Rm, ma_sp, label='Sweet-Parker: M_A \sim R_m^{-1/2}', color='green')
plt.loglog(Rm, ma_petschek, label='Petschek: M_A \sim \frac{\pi}{8 \ln R_m}', color='red')

# 그래프 스타일링
plt.xlabel('R_m (Magnetic Reynolds number)')
plt.ylabel('M_A (Reconnection rate)')
plt.title('Comparison of Reconnection Models')
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.legend()
plt.ylim(0.005, 0.5) # 원본 이미지와 유사한 범위
plt.xlim(100, 10**8)

plt.show()
```
